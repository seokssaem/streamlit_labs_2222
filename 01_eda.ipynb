```python
import pandas as pd

data = {
    "height": [165, 172, None, 180, 158, 175],
    "weight": [55, 70, 65, 80, 50, 68],
    "age": [25, 30, 28, 35, 22, 29],
    "gender": ["F", "M", "M", "M", "F", "M"],
}

df = pd.DataFrame(data)

print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# 결측치 확인
print(df.isnull().sum())

# 방법 1: 결측치가 있는 행 제거
df_drop = df.dropna()

# 방법 2: 숫자 결측치를 평균값으로 채우기
df["height"] = df["height"].fillna(df["height"].mean())

# 방법 3: 범주형 결측치를 최빈값으로 채우기
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
```

```python
df["gender_num"] = df["gender"].map({"F": 0, "M": 1})
```

```python
df_encoded = pd.get_dummies(df, columns=["gender"])
print(df_encoded)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[["height_scaled"]] = scaler.fit_transform(df[["height"]])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[["height_standard"]] = scaler.fit_transform(df[["height"]])
```
